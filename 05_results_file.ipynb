```python
from google.colab import drive, files
drive.mount('/content/drive')
import json, os, pandas as pd

BASE = '/content/drive/MyDrive/longevity_project'
scores = pd.read_csv(f'{BASE}/results/senescence_baseline_pilot.csv')

reviews = {}
for fn in sorted(os.listdir(f'{BASE}/results/reviews')):
    reviews[fn.removesuffix('.json')] = json.load(open(f'{BASE}/results/reviews/{fn}'))

results = {'hallmark': 'senescence', 'cell_type': 'VCAP',
           'scores': scores[['rank', 'name', 'score', 'nes', 'pert_dose', 'pert_time']].to_dict('records'),
           'reviews': reviews}
path = f'{BASE}/results/senescence_results.json'
json.dump(results, open(path, 'w'), indent=2)
print(len(results['scores']), 'scores,', len(reviews), 'reviews')
files.download(path)
```
